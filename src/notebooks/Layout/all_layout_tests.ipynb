# Plot layout test

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
from z2hex.geometry import HeavyHexLattice

plaquette_width = 3
plaquette_height = 5

lattice = HeavyHexLattice(plaquette_width, plaquette_height)
lattice.plot_lattice(number_qubits=True)
```

```python
lattice.initial_qubit_layout(None, "ibm_marrakesh")
```

```python
all_layouts = lattice.get_all_layouts("ibm_marrakesh")
len(all_layouts)
```

```python
from z2hex.geometry import plot_layout_over_device

for i, layout in enumerate(all_layouts):
    plot_layout_over_device("ibm_marrakesh", layout)
    plt.savefig(f"layout_{i}.png")
```
