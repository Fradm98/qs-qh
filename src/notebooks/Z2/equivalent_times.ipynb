```python
J = 2.5
h = 0.1
lamb = 1
g = 10

final_time = 8
```

```python
from z2hex.geometry import HeavyHexLattice

plaquette_width = 3
plaquette_height = 3

lattice = HeavyHexLattice(plaquette_width, plaquette_height)
lattice.plot_lattice()
```

```python
from z2hex.circs import trotter_dt
import numpy as np

parameters = [J, h, lamb]
error = 0.1

print(trotter_dt(error, lattice, *parameters))

for i in range(len(parameters)):
    this_parameters = np.array(parameters) / parameters[i]
    print(trotter_dt(error, lattice, *this_parameters))
```
