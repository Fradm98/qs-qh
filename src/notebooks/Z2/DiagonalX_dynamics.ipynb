```python
from sympy.physics.paulialgebra import Pauli
from utils.paulis import PauliTensor
from sympy.core.numbers import One
import sympy as sp
```

```python
t = sp.symbols("t")
```

#### Gauge qubits

```python
pauli_factors = [Pauli(1) for i in range(3)]
identities = [One for i in range(3)]
middle_pauli_factors = [Pauli(3) if i == 1 else One for i in range(3)]
```

```python
f1 = PauliTensor(identities, sp.cos(t)) + PauliTensor(pauli_factors, 1j*sp.sin(t))
midf = PauliTensor(middle_pauli_factors)
f2 = PauliTensor(identities, sp.cos(t)) + PauliTensor(pauli_factors, -1j*sp.sin(t))

f1*midf*f2
```

#### Matter conected to two gauges

```python
left_pauli_factors = [Pauli(1) if i < 3 else One for i in range(5)]
right_pauli_factors = [Pauli(1) if i > 1 else One for i in range(5)]
middle_pauli_factors = [Pauli(3) if i == 2 else One for i in range(5)]
identities = [One]*5
```

```python
f1 = PauliTensor(identities, sp.cos(t)) + PauliTensor(left_pauli_factors, 1j*sp.sin(t)) 
f2 = PauliTensor(identities, sp.cos(t)) + PauliTensor(right_pauli_factors, 1j*sp.sin(t))
midf = PauliTensor(middle_pauli_factors)
f3 = PauliTensor(identities, sp.cos(t)) - PauliTensor(right_pauli_factors, 1j*sp.sin(t))
f4 = PauliTensor(identities, sp.cos(t)) - PauliTensor(left_pauli_factors, 1j*sp.sin(t))

res = f1*f2*midf*f3*f4
res
```

#### Matter connected to three gauges

```python
u02_factors = [Pauli(1) if i < 3 else One for i in range(7)]
u24_factors = [Pauli(1) if 1 < i < 5 else One for i in range(7)]
u26_factors = [Pauli(1) if i == 2 or i > 4 else One for i in range(7)]
middle_pauli_factors = [Pauli(3) if i == 2 else One for i in range(7)]
identities = [One]*7
```

```python
u02dg = PauliTensor(identities, sp.cos(t)) + PauliTensor(u02_factors, 1j*sp.sin(t))
u24dg = PauliTensor(identities, sp.cos(t)) + PauliTensor(u24_factors, 1j*sp.sin(t))
u26dg = PauliTensor(identities, sp.cos(t)) + PauliTensor(u26_factors, 1j*sp.sin(t))
midf = PauliTensor(middle_pauli_factors)
u26 = PauliTensor(identities, sp.cos(t)) - PauliTensor(u26_factors, 1j*sp.sin(t))
u24 = PauliTensor(identities, sp.cos(t)) - PauliTensor(u24_factors, 1j*sp.sin(t))
u02 = PauliTensor(identities, sp.cos(t)) - PauliTensor(u02_factors, 1j*sp.sin(t))

res = u02dg*u24dg*u26dg*midf*u26*u24*u02
res
```

```python
sp.simplify(res.summands[0].coefficient)
```
