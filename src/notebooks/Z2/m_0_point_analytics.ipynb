```python
from utils.paulis import Pauli, PauliTensor, PauliTensorSum
from sympy.core.numbers import One
from sympy import symbols, cos, sin, I, sqrt
```

We must compute the Heisemberg picture evolution of the local $\tau_n^z$. To do this, in the $m = 0$ case, notice that the Hamiltonian can be grouped in commuting terms corresponding to each of the links in the lattice.

Consider only nodes connected to three links because that is what is plotted

# Matter operators

```python
g, lamb, t = symbols(r"g λ t")

N = 7

link_paulis_identity = [One]*N

link_left_int_paulis = link_paulis_identity.copy()
link_left_int_paulis[0] = Pauli(1)
link_left_int_paulis[1] = Pauli(1)
link_left_int_paulis[2] = Pauli(1)
link_left_single_paulis = link_paulis_identity.copy()
link_left_single_paulis[1] = Pauli(3)

link_right_int_paulis = link_paulis_identity.copy()
link_right_int_paulis[2] = Pauli(1)
link_right_int_paulis[3] = Pauli(1)
link_right_int_paulis[4] = Pauli(1)
link_right_single_paulis = link_paulis_identity.copy()
link_right_single_paulis[3] = Pauli(3)

link_down_int_paulis = link_paulis_identity.copy()
link_down_int_paulis[2] = Pauli(1)
link_down_int_paulis[5] = Pauli(1)
link_down_int_paulis[6] = Pauli(1)
link_down_single_paulis = link_paulis_identity.copy()
link_down_single_paulis[5] = Pauli(3)

observable_paulis = link_paulis_identity.copy()
observable_paulis[2] = Pauli(3)
observable = PauliTensorSum(*[PauliTensor(observable_paulis, coefficient=1)])

angle = t*sqrt(g**2 + lamb**2)
int_coeff = lamb/sqrt(g**2 + lamb**2)
single_coeff = g/sqrt(g**2 + lamb**2)

link1exp = PauliTensorSum(*[PauliTensor([One]*N, coefficient=cos(angle)), PauliTensor(link_left_int_paulis, coefficient=I*int_coeff*sin(angle)), PauliTensor(link_left_single_paulis, coefficient=I*single_coeff*sin(angle))])
link2exp = PauliTensorSum(*[PauliTensor([One]*N, coefficient=cos(angle)), PauliTensor(link_right_int_paulis, coefficient=I*int_coeff*sin(angle)), PauliTensor(link_right_single_paulis, coefficient=I*single_coeff*sin(angle))])
link3exp = PauliTensorSum(*[PauliTensor([One]*N, coefficient=cos(angle)), PauliTensor(link_down_int_paulis, coefficient=I*int_coeff*sin(angle)), PauliTensor(link_down_single_paulis, coefficient=I*single_coeff*sin(angle))])

link1exp_dg = PauliTensorSum(*[PauliTensor([One]*N, coefficient=cos(angle)), PauliTensor(link_left_int_paulis, coefficient=-I*int_coeff*sin(angle)), PauliTensor(link_left_single_paulis, coefficient=-I*single_coeff*sin(angle))])
link2exp_dg = PauliTensorSum(*[PauliTensor([One]*N, coefficient=cos(angle)), PauliTensor(link_right_int_paulis, coefficient=-I*int_coeff*sin(angle)), PauliTensor(link_right_single_paulis, coefficient=-I*single_coeff*sin(angle))])
link3exp_dg = PauliTensorSum(*[PauliTensor([One]*N, coefficient=cos(angle)), PauliTensor(link_down_int_paulis, coefficient=-I*int_coeff*sin(angle)), PauliTensor(link_down_single_paulis, coefficient=-I*single_coeff*sin(angle))])

p = link1exp_dg * link2exp_dg * link3exp_dg * observable * link3exp * link2exp * link1exp
# p = link2exp_dg * observable * link2exp
```

```python
p.simplify().summands
```

```python
from sympy import simplify

simplify(p.summands[0].coefficient)
```

```python
g, lamb, t = symbols(r"g λ t")

N = 7

link_paulis_identity = [One]*N

link_left_int_paulis = link_paulis_identity.copy()
link_left_int_paulis[0] = Pauli(1)
link_left_int_paulis[1] = Pauli(1)
link_left_int_paulis[2] = Pauli(1)
link_left_single_paulis = link_paulis_identity.copy()
link_left_single_paulis[1] = Pauli(3)

link_right_int_paulis = link_paulis_identity.copy()
link_right_int_paulis[2] = Pauli(1)
link_right_int_paulis[3] = Pauli(1)
link_right_int_paulis[4] = Pauli(1)
link_right_single_paulis = link_paulis_identity.copy()
link_right_single_paulis[3] = Pauli(3)

link_down_int_paulis = link_paulis_identity.copy()
link_down_int_paulis[2] = Pauli(1)
link_down_int_paulis[5] = Pauli(1)
link_down_int_paulis[6] = Pauli(1)
link_down_single_paulis = link_paulis_identity.copy()
link_down_single_paulis[5] = Pauli(3)

observable_paulis = link_paulis_identity.copy()
observable_paulis[3] = Pauli(1)
observable = PauliTensorSum(*[PauliTensor(observable_paulis, coefficient=1)])

angle = t*sqrt(g**2 + lamb**2)
int_coeff = lamb/sqrt(g**2 + lamb**2)
single_coeff = g/sqrt(g**2 + lamb**2)

link1exp = PauliTensorSum(*[PauliTensor([One]*N, coefficient=cos(angle)), PauliTensor(link_left_int_paulis, coefficient=I*int_coeff*sin(angle)), PauliTensor(link_left_single_paulis, coefficient=I*single_coeff*sin(angle))])
link2exp = PauliTensorSum(*[PauliTensor([One]*N, coefficient=cos(angle)), PauliTensor(link_right_int_paulis, coefficient=I*int_coeff*sin(angle)), PauliTensor(link_right_single_paulis, coefficient=I*single_coeff*sin(angle))])
link3exp = PauliTensorSum(*[PauliTensor([One]*N, coefficient=cos(angle)), PauliTensor(link_down_int_paulis, coefficient=I*int_coeff*sin(angle)), PauliTensor(link_down_single_paulis, coefficient=I*single_coeff*sin(angle))])

link1exp_dg = PauliTensorSum(*[PauliTensor([One]*N, coefficient=cos(angle)), PauliTensor(link_left_int_paulis, coefficient=-I*int_coeff*sin(angle)), PauliTensor(link_left_single_paulis, coefficient=-I*single_coeff*sin(angle))])
link2exp_dg = PauliTensorSum(*[PauliTensor([One]*N, coefficient=cos(angle)), PauliTensor(link_right_int_paulis, coefficient=-I*int_coeff*sin(angle)), PauliTensor(link_right_single_paulis, coefficient=-I*single_coeff*sin(angle))])
link3exp_dg = PauliTensorSum(*[PauliTensor([One]*N, coefficient=cos(angle)), PauliTensor(link_down_int_paulis, coefficient=-I*int_coeff*sin(angle)), PauliTensor(link_down_single_paulis, coefficient=-I*single_coeff*sin(angle))])

p = link1exp_dg * link2exp_dg * link3exp_dg * observable * link3exp * link2exp * link1exp
# p = link2exp_dg * observable * link2exp
```

```python
p.simplify()
```

# Gauge operators

```python
g, lamb, t = symbols(r"g λ t")

N = 3

link_paulis_identity = [One]*N

link_int_paulis = link_paulis_identity.copy()
link_int_paulis[0] = Pauli(1)
link_int_paulis[1] = Pauli(1)
link_int_paulis[2] = Pauli(1)
link_single_paulis = link_paulis_identity.copy()
link_single_paulis[1] = Pauli(3)

observable_paulis = link_paulis_identity.copy()
observable_paulis[1] = Pauli(3)
observable = PauliTensorSum(*[PauliTensor(observable_paulis, coefficient=1)])

angle = t*sqrt(g**2 + lamb**2)
int_coeff = lamb/sqrt(g**2 + lamb**2)
single_coeff = g/sqrt(g**2 + lamb**2)

linkexp = PauliTensorSum(*[PauliTensor([One]*N, coefficient=cos(angle)), PauliTensor(link_int_paulis, coefficient=I*int_coeff*sin(angle)), PauliTensor(link_single_paulis, coefficient=I*single_coeff*sin(angle))])

linkexp_dg = PauliTensorSum(*[PauliTensor([One]*N, coefficient=cos(angle)), PauliTensor(link_int_paulis, coefficient=-I*int_coeff*sin(angle)), PauliTensor(link_single_paulis, coefficient=-I*single_coeff*sin(angle))])

p = linkexp_dg * observable * linkexp
# p = link2exp_dg * observable * link2exp
```

```python
p.simplify().summands
```
