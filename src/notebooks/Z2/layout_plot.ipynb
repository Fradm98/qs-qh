## Mapping

```python
logical_to_physical_list = list(range(3, 20)) + \
                      list(range(21, 40)) + \
                      list(range(41, 60)) + \
                      list(range(61, 80)) + \
                      list(range(81, 100)) + \
                      list(range(101, 120)) + \
                      list(range(121, 140)) + \
                      list(range(143, 156))

logical_to_physical_map = {i: j for i, j in zip(range(len(logical_to_physical_list)), logical_to_physical_list)}
physical_to_logical_map = {v:k for k, v in logical_to_physical_map.items()}
```
